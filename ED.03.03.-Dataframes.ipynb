### Creación de DataFrames con Spark

**1. Desde una lista**

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("DataFrame-Demo").getOrCreate()

data = [("Aitor", 33), ("Maria", 28), ("Juan", 45)]
columns = ["Nombre", "Edad"]

# ✅ OK: crear DataFrame directamente
df = spark.createDataFrame(data, columns)
df.show()
```

**2. Desde un RDD**

```python
# Crear un RDD de tuplas
rdd = sc.parallelize([("Angel", 9), ("Maria", 3), ("Ramon", 7)])

# Convertir a DataFrame (Lazy)
df = rdd.toDF(["nombre", "nota"])

# Acción (Eager, carga los datos)
df.show()
```

**3. Leyendo ficheros**

```python
# Leyendo un CSV
df_csv = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("datos/ventas.csv")

# ❌ INCORRECTO: no especificar opciones importantes como header en CSV
# df_malo = spark.read.csv("datos/ventas.csv")

# Leyendo un Parquet (recomendado)
df_parquet = spark.read.parquet("datos/ventas.parquet")

# Lectura con opciones extra (separador personalizado, etc.)
df2 = spark.read.csv(
    "ventas.csv",
    header=True,        # La primera línea son los nombres de columna
    inferSchema=True,   # Spark adivina los tipos de datos (int, string...)
    sep=";"              # Separador personalizado
)
df2.show(5)
```
